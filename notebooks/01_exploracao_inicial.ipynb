```python
# Vamos calcular as porcentagens#

percentual_ausentes_shades = (
    df_shades.isna().mean() * 100
).round(2)

percentual_ausentes_numbers = (
    df_numbers.isna().mean() * 100
).round(2)

print("=== PERCENTUAL DE AUSENTES: ALL SHADES ===")
print(
    percentual_ausentes_shades[
        percentual_ausentes_shades > 0
    ]
)

print("\n=== PERCENTUAL DE AUSENTES: ALL NUMBERS ===")
print(
    percentual_ausentes_numbers[
        percentual_ausentes_numbers > 0
    ]
)
```

```python
#USAMOS PARA COMPARAR SE HÁ DUPLICIDADE#

duplicadas_shades = df_shades.duplicated().sum()
duplicadas_numbers = df_numbers.duplicated().sum()

print("Linhas totalmente duplicadas em allShades:", duplicadas_shades)
print("Linhas totalmente duplicadas em allNumbers:", duplicadas_numbers)
```

```python
print("=== LUMINOSIDADE: ALL SHADES ===")
print("Menor valor:", df_shades["lightness"].min())
print("Maior valor:", df_shades["lightness"].max())
print("Média:", df_shades["lightness"].mean())

print("\n=== LUMINOSIDADE: ALL NUMBERS ===")
print("Menor valor:", df_numbers["lightness"].min())
print("Maior valor:", df_numbers["lightness"].max())
print("Média:", df_numbers["lightness"].mean())
```

```python
#ESTAMOS CRIANDO CÓPIAS PARA QUE CASO DÊ ALGUM PORBELMA AINDA TEREMOS OS DADOS ORIGINAIS DENTRO DO NOTEBOOK#
df_shades_limpo = df_shades.copy()
df_numbers_limpo = df_numbers.copy()
```

```python
#VAMOS CRIAR A LUMINOSIDADE NA ESCALA DE 0 A 100#

df_shades_limpo["lightness_100"] = (
    df_shades_limpo["lightness"] * 100
)

df_numbers_limpo["lightness_100"] = (
    df_numbers_limpo["lightness"] * 100
)
```

```python
#AQUI NÓS ESTAMOS ARREDONDANDO PARA APENAS DUAS CASA DECIMAIS - VÊ SE NÃO ESQUECE#

df_shades_limpo["lightness_100"] = (
    df_shades_limpo["lightness_100"].round(2)
)

df_numbers_limpo["lightness_100"] = (
    df_numbers_limpo["lightness_100"].round(2)
)
```

```python
df_shades_limpo[
    ["brand", "product", "hex", "lightness", "lightness_100"]
].head()
```

```python
print("=== ALL SHADES — ESCALA 0 A 100 ===")
print("Mínimo:", df_shades_limpo["lightness_100"].min())
print("Máximo:", df_shades_limpo["lightness_100"].max())
print("Média:", df_shades_limpo["lightness_100"].mean().round(2))

print("\n=== ALL NUMBERS — ESCALA 0 A 100 ===")
print("Mínimo:", df_numbers_limpo["lightness_100"].min())
print("Máximo:", df_numbers_limpo["lightness_100"].max())
print("Média:", df_numbers_limpo["lightness_100"].mean().round(2))
```

```python
#NESSA PARTE VAMOS TRANSFORMAR A LUMINOSIDADE NUMÉRICA EM FAIXAS DE TONS - ACREDITO EU SER UMA DAS PARTES MAIS IMPORTANTES, BUT HO AM I?#

faixas_luminosidade = [0, 30, 50, 75, 100]

categorias_tons = [
    "Retinto / Deep",
    "Escuro / Dark",
    "Médio / Medium",
    "Claro / Light"
]

df_shades_limpo["categoria_tom"] = pd.cut(
    df_shades_limpo["lightness_100"],
    bins=faixas_luminosidade,
    labels=categorias_tons,
    include_lowest=True
)

df_numbers_limpo["categoria_tom"] = pd.cut(
    df_numbers_limpo["lightness_100"],
    bins=faixas_luminosidade,
    labels=categorias_tons,
    include_lowest=True
)
```

```python
#TABELA PRONTA - PELO VISTO VC NÃO É TÃO LERDO ASSIM#

df_shades_limpo[
    [
        "brand",
        "product",
        "hex",
        "lightness_100",
        "categoria_tom"
    ]
].head(10)
```

```python
# AQUI VAMOS VERIFICAR SE ALGUM REGISTRO FICOU SEM CATEGORIA#

print(
    "Sem categoria em allShades:",
    df_shades_limpo["categoria_tom"].isna().sum()
)

print(
    "Sem categoria em allNumbers:",
    df_numbers_limpo["categoria_tom"].isna().sum()
)
```

```python
# ESSA PARTE NÓS ESTAMOS LISTANDO OS TONS POR CATEGORIA#

contagem_categorias = (
    df_shades_limpo["categoria_tom"]
    .value_counts(sort=False)
)

print("Quantidade de registros por categoria:")
print(contagem_categorias)
```

```python
#Por que usar percentual? - A quantidade absoluta informa quantos registros existem. Já o percentual mostra quanto cada categoria representa da base inteira.#

percentual_categorias = (
    df_shades_limpo["categoria_tom"]
    .value_counts(normalize=True, sort=False)
    .mul(100)
    .round(2)
)

print("Percentual de registros por categoria:")
print(percentual_categorias)
```

```python
resumo_categorias = pd.DataFrame({
    "quantidade": contagem_categorias,
    "percentual": percentual_categorias
}).reset_index()

resumo_categorias.columns = [
    "categoria_tom",
    "quantidade",
    "percentual"
]

resumo_categorias
```

```python
#CRIANDO NOSSO PRIMEIRO GRÁFICO#

plt.figure(figsize=(10, 5))

plt.bar(
    resumo_categorias["categoria_tom"],
    resumo_categorias["quantidade"]
)

plt.title("Distribuição de tonalidades por categoria")
plt.xlabel("Categoria de tonalidade")
plt.ylabel("Quantidade de tons")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# COM O PLT-.VAR - ESTAMOPS ESCREVENDO A QUANTIDADE CORRESPONDETE ACIMA DE CADA BARRA - UMA FORMA DE TER MAIS UMA INFORMAÇÃO IMPORTANTE#

plt.figure(figsize=(10, 5))

barras = plt.bar(
    resumo_categorias["categoria_tom"],
    resumo_categorias["quantidade"]
)

plt.title("Distribuição de tonalidades por categoria")
plt.xlabel("Categoria de tonalidade")
plt.ylabel("Quantidade de tons")

plt.bar_label(barras)

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
