Unificação das informações de identificação do tom

A análise mostrou que não existe nenhum registro com nome_tom e tom_especifico simultaneamente nulos. Portanto, todos os registros possuem pelo menos uma informação de identificação do tom.

Diante disso, foi avaliada a possibilidade de unificar essas informações.

As colunas possuem características diferentes: nome_tom contém nomes comerciais ou criativos atribuídos pelas marcas, sem uma padronização consistente entre os produtos, enquanto tom_especifico apresenta informações mais recorrentes e estruturadas para identificação do tom.

Como o objetivo das análises posteriores é trabalhar com uma única informação de identificação do tom, optei por criar a coluna tom_unificado. Nessa nova coluna, será priorizado o valor de tom_especifico e, nos casos em que ele estiver ausente, será utilizado o valor disponível em nome_tom.

Dessa forma, preservo a informação disponível nos dados, reduzo a presença de valores nulos na variável que poderá ser utilizada para análise e mantenho as duas colunas originais para rastreabilidade da transformação.

```python
# Cria uma coluna unificada aproveitando o nome_tom onde tom_especifico for NaN
df_shades["tom_unificado"] = df_shades["tom_especifico"].fillna(df_shades["nome_tom"])
print(df_shades[["nome_tom", "tom_especifico", "tom_unificado"]])
```

PADRONIZAÇÃO DO DATAFRAME PRA MELHOR UTILIZAÇÃO DELES NO SQL

```python
# ==============================================================================
# ETAPA 1: Padronização de Texto (Para facilitar as consultas no SQL)
# ==============================================================================

# Lista com todas as colunas que contêm texto (excluímos matiz, saturacao e luminosidade por serem números)
colunas_texto = [
    'marca', 'produto', 'url_produto', 'url_imagem', 'descricao', 
    'nome_tom', 'tom_especifico', 'espaco_cores', 'codigo_hex','tom_unificado'
]

# O acessor .str do Pandas aplica a limpeza mantendo os valores nulos (NaN) intactos!
for col in colunas_texto:
    # 1. .str.strip() -> Remove espaços no início e no final do texto (ex: " tom " vira "tom")
    # 2. .str.lower() -> Transforma todo o texto em minúsculo (ex: "Ruby" vira "ruby")
    df_shades[col] = df_shades[col].str.strip().str.lower()
```

```python
df_shades.head(10)
```

```python
# Cria uma coluna de ID numérico sequencial (1, 2, 3...)
df_shades['id_shade'] = range(1, len(df_shades) + 1)

# Reorganiza as colunas para colocar o 'id_shade' como a primeira coluna da tabela
colunas = ['id_shade'] + [col for col in df_shades.columns if col != 'id_shade']
df_shades = df_shades[colunas]
```

```python
df_shades.info()
```

```python
print(df_shades['luminosidade'].describe())
```

```python
# ------------------------------------------------------------------------------
# ADEQUAÇÃO DA ESCALA DE LUMINOSIDADE (DE DECIMAL PARA PERCENTUAL 0-100)
# ------------------------------------------------------------------------------
# A variável 'luminosidade' varia de 0.0 a 1.0. Multiplicamos por 100 para 
# atender à regra de negócio (faixas de 0 a 100) especificada no case.
df_shades['luminosidade'] = df_shades['luminosidade'] * 100
```

```python
# ------------------------------------------------------------------------------
# 3. CRIAÇÃO DAS FAIXAS (BINS) E RÓTULOS (MISSÃO 1 DO CASE)
# ------------------------------------------------------------------------------
# Regra do Desafio:
#  0 a 30  -> Retinto / Deep
# 31 a 50  -> Escuro / Dark
# 51 a 75  -> Médio / Medium
# 76 a 100 -> Claro / Light

# 1. BINS (PONTOS DE CORTE / FAiXAS NUMÉRICAS):
# Definimos os limites numéricos para fatiar a coluna 'luminosidade'.
# Nota: Usamos -1 ao invés de 0 no início para garantir que o valor '0' 
# seja incluído na primeira faixa (de 0 a 30).
# As 4 faixas criadas são: (-1 a 30], (30 a 50], (50 a 75], (75 a 100]
bins = [-1, 30, 50, 75, 100]
# 2. LABELS (ETIQUETAS DE TEXTO / RÓTULOS):
# Definimos os nomes que cada uma das 4 faixas numéricas vai receber.
# A ordem aqui DEVE corresponder perfeitamente à ordem das faixas dos bins.
labels = ['Retinto / Deep', 'Escuro / Dark', 'Médio / Medium', 'Claro / Light']

# 3. AGRUPAMENTO COM PD.CUT (CORTANDO OS DADOS E CRIANDO A NOVA COLUNA):
# O pd.cut pega a coluna 'luminosidade', compara cada número com as faixas em 'bins',
# e substitui pelo nome correspondente presente em 'labels'.
# O resultado é guardado na NOVA coluna chamada 'grupo_de_tom'.
df_shades['grupo_de_tom'] = pd.cut(
    df_shades['luminosidade'], # Coluna original com os números (0 a 100)
    bins=bins, # Limites numéricos onde o corte é feito
    labels=labels # Nomes em texto que serão atribuídos a cada faixa
)

# Visualização do resultado final: Exibe as primeiras 10 linhas das colunas 'luminosidade' e 'grupo_de_tom' para verificar se a categorização foi aplicada corretamente.
print(df_shades[['luminosidade', 'grupo_de_tom']].head(555))
```

```python
# 4. VALIDAÇÃO DO RESULTADO DA CATEGORIZAÇÃO
print("\n--- Distribuição das Bases por Categoria de Tom (Visão do Mercado) ---")
print(df_shades['grupo_de_tom'].value_counts(dropna=False))

print("\n--- Percentual do Mercado por Categoria ---")
print((df_shades['grupo_de_tom'].value_counts(normalize=True) * 100).round(2).astype(str) + '%')
```
