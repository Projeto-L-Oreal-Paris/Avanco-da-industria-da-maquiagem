```python
#==============================================================================
#ETAPA 5 - ANÁLISE DA ESTRUTURA DAS TABELAS
#Objetivo:
#Compreender como os DataFrames se relacionam antes de realizar
#novas transformações. Nesta etapa investigamos se as tabelas
#possuem colunas em comum, possíveis chaves e se faz sentido
#mantê-las separadas ou unificá-las.
#==============================================================================
```

```python
# ETAPA 5.1 - DIMENSÃO DAS TABELAS
#
# Objetivo:
# Verificar a quantidade de linhas e colunas de cada DataFrame.
# Essa análise ajuda a identificar se ambas possuem estrutura
# semelhante, um dos indícios para um possível relacionamento.

print("Dimensão do allShades:")
print(df_shades.shape)

print("\nDimensão do allNumbers:")
print(df_numbers.shape)
# As tabelas não representam exatamente o mesmo conjunto de observações.
```

```python
# ==============================================================================
# ETAPA 5.2 - LISTAGEM DAS COLUNAS
#
# Objetivo:
# Exibir todas as colunas presentes em cada tabela para
# identificar possíveis semelhanças e diferenças.
# ==============================================================================

print("Colunas do allShades:\n")
print(df_shades.columns.tolist())

print("\n-----------------------------\n")

print("Colunas do allNumbers:\n")
print(df_numbers.columns.tolist())
```

```python
# ==============================================================================
# ETAPA 5.3 - COLUNAS EM COMUM
#
# Objetivo:
# Identificar quais colunas existem simultaneamente nos dois DataFrames.
# ==============================================================================

colunas_comuns = sorted(
    set(df_shades.columns).intersection(df_numbers.columns)
)

print("Colunas em comum:")

for coluna in colunas_comuns:
    print(f"- {coluna}")

# Isso significa que elas falam sobre o mesmo domínio (tons de maquiagem).
# Mas isso não significa que cada linha de uma corresponde a uma linha da outra.
# É uma diferença importante.
```

```python
# ==============================================================================
# ETAPA 5.4 - COLUNAS EXCLUSIVAS
#
# Objetivo:
# Identificar quais colunas são exclusivas de cada DataFrame.
# ==============================================================================

colunas_shades = sorted(
    set(df_shades.columns) - set(df_numbers.columns)
)

colunas_numbers = sorted(
    set(df_numbers.columns) - set(df_shades.columns)
)

print("Colunas exclusivas do allShades:")

for coluna in colunas_shades:
    print(f"- {coluna}")

print("\n-----------------------------\n")

print("Colunas exclusivas do allNumbers:")

for coluna in colunas_numbers:
    print(f"- {coluna}")
```

```python
# ==============================================================================
# ETAPA 5.5 - VERIFICAÇÃO DE CHAVE NATURAL
#
# Objetivo:
# Verificar se a combinação inicial das colunas 'marca' + 'produto' + 'nome_tom'
# é suficiente para identificar cada registro de forma única em cada DataFrame.
# ==============================================================================

duplicados_shades = df_shades.duplicated(
    subset=["marca", "produto", "nome_tom"]
).sum()

duplicados_numbers = df_numbers.duplicated(
    subset=["marca", "produto", "nome_tom"]
).sum()

print(f"Duplicados (allShades): {duplicados_shades}")
print(f"Duplicados (allNumbers): {duplicados_numbers}")

# Conclusão desta célula: Como houve milhares de duplicados, essa combinação 
# NÃO é uma chave primária válida. Precisamos investigar chaves mais específicas.
```

```python
# ==============================================================================
# ETAPA 5.6 - TESTE DE CHAVE COMPOSTA COM 'tom_especifico'
#
# Objetivo:
# Testar se a inclusão da coluna 'tom_especifico' na chave composta 
# (marca + produto + tom_especifico) consegue conectar os DataFrames 
# sem gerar duplicação de linhas.
# ==============================================================================

teste_merge = df_numbers.merge(
    df_shades,
    on=["marca", "produto", "tom_especifico"],
    how="left",
    indicator=True
)

print("Resultado da verificação de chave composta:")
print(teste_merge["_merge"].value_counts())

# Conclusão desta célula: O resultado deu 3.117 no 'both' sem nenhuma duplicação!
# Isso confirma que 'marca + produto + tom_especifico' é a chave perfeita.
```

```python
# ==============================================================================
# ETAPA 5.6.1 - TESTE DE CHAVE COMPOSTA
#
# Objetivo:
# Verificar se a inclusão da coluna 'codigo_hex' reduz
# as múltiplas correspondências encontradas no Merge.
#
# Hipótese:
# A combinação
# marca + produto + nome_tom + codigo_hex
# pode representar melhor um registro único.
# ==============================================================================

teste_merge = df_numbers.merge(
    df_shades,
    on=[
        "marca",
        "produto",
        "nome_tom",
        "codigo_hex"
    ],
    how="left",
    indicator=True
)

print(teste_merge["_merge"].value_counts())
```

```python
# ==============================================================================
# ETAPA 5.6.2 - TESTE DE CHAVE COMPOSTA
#
# Objetivo:
# Avaliar se a luminosidade pode ser utilizada como parte
# da chave de relacionamento entre as tabelas.
# ==============================================================================

teste_merge = df_numbers.merge(
    df_shades,
    on=[
        "marca",
        "produto",
        "luminosidade"
    ],
    how="left",
    indicator=True
)

print(teste_merge["_merge"].value_counts())
```

```python
# ==============================================================================
# ETAPA 5.6.3 - TESTE DE CHAVE COMPOSTA COM LUMINOSIDADE E CÓDIGO HEX
#
# Objetivo:
# Investigar se a combinação entre 'marca', 'produto', 'luminosidade' e 'codigo_hex'
# reduz as correspondências duplicadas observadas no teste anterior.
# ==============================================================================

teste_merge = df_numbers.merge(
    df_shades,
    on=[
        "marca",
        "produto",
        "luminosidade",
        "codigo_hex"
    ],
    how="left",
    indicator=True
)

print("Resultado do teste com Luminosidade + Código HEX:")
print(teste_merge["_merge"].value_counts())

# Conclusão desta célula: Retornou 3.192 linhas. Reduziu bastante, mas ainda 
# gerou algumas duplicidades residuais (já que a allNumbers tem 3.117 linhas).
```

```python
# ==============================================================================
# ETAPA 5.6.4 - TESTE DE CHAVE COMPOSTA COM 'tom_especifico'
#
# Objetivo:
# Avaliar se a coluna 'tom_especifico' (junto de 'marca' e 'produto') 
# representa a combinação perfeita para identificar unicamente cada registro.
# ==============================================================================

teste_merge = df_numbers.merge(
    df_shades,
    on=[
        "marca",
        "produto",
        "tom_especifico"
    ],
    how="left",
    indicator=True
)

print("Resultado do teste com tom_especifico:")
print(teste_merge["_merge"].value_counts())

# Conclusão desta célula: Retornou exatamente 3.117 no 'both' sem nenhuma duplicação!
# Encontramos a chave primária composta ideal!
```

```python
# ==============================================================================
# ETAPA 5.6.5 - VALIDAÇÃO DA INTEGRIDADE DAS COLUNAS DUPLICADAS
#
# Objetivo:
# Confirmar se as colunas compartilhadas entre as tabelas (nome_tom, codigo_hex e 
# luminosidade) possuem valores 100% idênticos ao usar a chave composta encontrada,
# tratando adequadamente os valores nulos (NaN) para evitar falsos negativos no Python.
# ==============================================================================

# Merge temporário para teste de integridade
df_teste_validacao = df_numbers.merge(
    df_shades,
    on=["marca", "produto", "tom_especifico"],
    suffixes=("_numbers", "_shades")
)

# 1. Validação do Nome do tom (usando fillna("") para considerar nulo igual a nulo)
valida_nome = df_teste_validacao["nome_tom_numbers"].fillna("").equals(
    df_teste_validacao["nome_tom_shades"].fillna("")
)

# 2. Validação do Código HEX
valida_hex = (
    df_teste_validacao["codigo_hex_numbers"] == df_teste_validacao["codigo_hex_shades"]
).all()

# 3. Validação da Luminosidade
valida_lum = (
    df_teste_validacao["luminosidade_numbers"] == df_teste_validacao["luminosidade_shades"]
).all()

print(f"Nome do tom confere 100%? {valida_nome}")
print(f"Código HEX confere 100%? {valida_hex}")
print(f"Luminosidade confere 100%? {valida_lum}")
```

```python
# ==============================================================================
# ETAPA 5.7 - O QUE APRENDEMOS SOBRE AS DUAS TABELAS? (CONCLUSÃO DIDÁTICA)
#
# Depois de investigar e testar os dados, chegamos a 4 conclusões principais:
#
# 1. O TAMANHO DAS TABELAS (Quem é quem?):
#    • allShades (6.816 linhas): É a nossa base principal. Ela tem TODAS as 
#      maquiagens pesquisadas no mercado.
#    • allNumbers (3.117 linhas): É um "pedaço" da base principal. Ela só 
#      contém as maquiagens que usam números na nomenclatura (ex: tom 10, 20, 30).
#
# 2. AS INFORMAÇÕES SE COMPLEMENTAM:
#    • A allShades traz links de fotos e detalhes de cor.
#    • A allNumbers traz a ordem numérica dos tons.
#    • Elas falam do mesmo assunto, mas cada uma guarda um pedaço da informação!
#
# 3. A COMBINAÇÃO PERFEITA DE CHAVE (Como conectar as duas?):
#    • Testamos juntar por 'marca + produto + nome_tom', mas deu duplicação.
#    • A combinação que deu 100% CERTO sem repetir linhas foi:
#      ==> marca + produto + tom_especifico <==
#
# 4. A DECISÃO DO NOSSO ETL (O que vamos fazer agora?):
#    • Para facilitar a nossa vida no Power BI, não precisamos deixar solto.
#    • Vamos fazer um LEFT JOIN partindo da allShades (base maior).
#    • Assim, mantemos as 6.816 bases do mercado e "encaixamos" os números 
#      nas maquiagens que têm essa informação, gerando UM ÚNICO ARQUIVO limpo.
# ==============================================================================
```

```python
# ==============================================================================
# ETAPA 5.8 - UNIFICAÇÃO FINAL DA BASE 
#
# Objetivo:
# Unificar os DataFrames partindo da base principal (df_shades) com um LEFT JOIN.
# Isso garante que mantemos todas as 6.816 tonalidades do mercado e adicionamos
# os dados de ordenação (números) apenas onde eles existem.
# ==============================================================================

# 1. Selecionamos apenas as colunas EXCLUSIVAS da df_numbers para não duplicar colunas
cols_exclusivas_numbers = ['marca', 'produto', 'tom_especifico', 'ordem_luminosidade', 'numero_tom', 'id_tom']

# 2. Fazemos o merge partindo de df_shades (Base Principal)
df_unificado = df_shades.merge(
    df_numbers[cols_exclusivas_numbers],
    on=["marca", "produto", "tom_especifico"],
    how="left"
)

# Validação do tamanho da base final:
print(f"Total de registros na base unificada: {len(df_unificado)}")
# SAÍDA ESPERADA: 6816 linhas (perfeito para a análise do mercado todo!)
```

```python
# ==============================================================================
# ETAPA 6.0 - ANÁLISE DE DADOS AUSENTES E CATEGORIZAÇÃO DA LUMINOSIDADE
#
# Objetivo:
# Avaliar a presença de valores ausentes oriundos da junção das tabelas, 
# documentar a decisão técnica de preservação dos dados e aplicar a 
# categorização de luminosidade especificada na Missão 1 do desafio.
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. ANÁLISE E DIAGNÓSTICO DE VALORES AUSENTES
# ------------------------------------------------------------------------------
print("--- Verificação de Valores Ausentes no Dataset Unificado ---")
print(df_unificado.isnull().sum())
# CONCLUSÃO DA ANÁLISE DE AUSENTES:
# Os valores ausentes em 'ordem_luminosidade', 'numero_tom' e 'id_tom' são 
# decorrentes da junção (LEFT JOIN) com a tabela df_numbers (3.117 registros). 
# DECISÃO TÉCNICA: Mantemos os valores nulos nessas colunas sem excluir linhas, 
# preservando o universo completo de 6.816 produtos para as análises de mercado.
```

```python
# ------------------------------------------------------------------------------
# 2. ADEQUAÇÃO DA ESCALA DE LUMINOSIDADE (DE DECIMAL PARA PERCENTUAL 0-100)
# ------------------------------------------------------------------------------
# A variável 'luminosidade' varia de 0.0 a 1.0. Multiplicamos por 100 para 
# atender à regra de negócio (faixas de 0 a 100) especificada no case.
df_unificado['luminosidade_pct'] = df_unificado['luminosidade'] * 100
df_unificado['luminosidade_pct'] = df_unificado['luminosidade'] * 100
```

```python
# ------------------------------------------------------------------------------
# 3. CRIAÇÃO DAS FAIXAS (BINS) E RÓTULOS (MISSÃO 1 DO CASE)
# ------------------------------------------------------------------------------
# Regra do Desafio:
#  0 a 30  -> Retinto / Deep
# 31 a 50  -> Escuro / Dark
# 51 a 75  -> Médio / Medium
# 76 a 100 -> Claro / Light
bins = [-1, 30, 50, 75, 100]
labels = ['Retinto / Deep', 'Escuro / Dark', 'Médio / Medium', 'Claro / Light']

df_unificado['grupo_de_tom'] = pd.cut(
    df_unificado['luminosidade_pct'], 
    bins=bins, 
    labels=labels
)
```

```python
# 4. VALIDAÇÃO DO RESULTADO DA CATEGORIZAÇÃO
print("\n--- Distribuição das Bases por Categoria de Tom (Visão do Mercado) ---")
print(df_unificado['grupo_de_tom'].value_counts(dropna=False))

print("\n--- Percentual do Mercado por Categoria ---")
print((df_unificado['grupo_de_tom'].value_counts(normalize=True) * 100).round(2).astype(str) + '%')
```

```python
# ==============================================================================
# ETAPA 7.0 - EXPORTAÇÃO DA BASE TRATADA PARA ANÁLISES NO POWER BI / DAX
#
# Objetivo:
# Salvar o dataset unificado em formato CSV com codificação utf-8-sig
# (compatível com acentuação no Power BI / Excel).
# ==============================================================================

df_unificado.to_csv('base_maquiagem_unificada.csv', index=False, encoding='utf-8-sig')

print("[OK] Arquivo 'base_maquiagem_unificada.csv' gerado com sucesso!")
print(f"Dimensões do arquivo entregue: {df_unificado.shape}") 
```
